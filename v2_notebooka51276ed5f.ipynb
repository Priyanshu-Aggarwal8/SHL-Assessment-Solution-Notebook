```python
# ============================================================
# SHL GRAMMAR SCORING ENGINE 2026
# Multimodal Audio + Grammar + Semantic Ensemble
# ============================================================
#
# Pipeline:
#
# WAV
#  │
#  ├── Whisper ASR
#  │      ├── transcript
#  │      ├── timing
#  │      ├── confidence
#  │      └── no-speech probability
#  │
#  ├── Acoustic / Prosody features
#  │      ├── RMS
#  │      ├── ZCR
#  │      ├── spectral features
#  │      ├── MFCC
#  │      ├── pitch
#  │      └── silence / pauses
#  │
#  └── Transcript
#         ├── LanguageTool grammar features
#         ├── lexical features
#         ├── MPNet embeddings
#         └── TF-IDF word + character ngrams
#
# Models:
#   1. ExtraTrees
#   2. LightGBM
#   3. MPNet + numeric Ridge
#   4. TF-IDF Ridge
#
# Final:
#   OOF-optimized convex ensemble
#
# ============================================================


# ============================================================
# 0. INSTALL REQUIRED PACKAGES
# ============================================================

!pip install -q \
    openai-whisper \
    sentence-transformers \
    language-tool-python \
    textstat \
    lightgbm


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import re
import json
import random
import warnings

from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import librosa
import soundfile as sf
import whisper
import torch

from tqdm.auto import tqdm

from sentence_transformers import SentenceTransformer
import language_tool_python
import textstat

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import Ridge
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import FeatureUnion
from sklearn.base import clone

from scipy.stats import pearsonr
from scipy.optimize import minimize

from lightgbm import LGBMRegressor


warnings.filterwarnings("ignore")

SEED = 2026

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))


# ============================================================
# 2. DOWNLOAD COMPETITION DATA
# ============================================================

import kagglehub

path = kagglehub.competition_download(
    "shl-hiring-assessment-2026"
)

print("Competition path:")
print(path)


# ============================================================
# 3. CONFIGURATION
# ============================================================

dataset_base = os.path.join(
    path,
    "Dataset_Final"
)

CONFIG = {

    "train_audio_dir":
        os.path.join(dataset_base, "train"),

    "test_audio_dir":
        os.path.join(dataset_base, "test"),

    "train_csv":
        os.path.join(dataset_base, "train.csv"),

    "test_csv":
        os.path.join(dataset_base, "test.csv"),

    "sample_submission":
        os.path.join(dataset_base, "sample_submission.csv"),

    "whisper_model":
        "small",

    "embedding_model":
        "sentence-transformers/all-mpnet-base-v2",

    "random_state":
        SEED
}


# ============================================================
# 4. LOAD CSV FILES
# ============================================================

train_df = pd.read_csv(
    CONFIG["train_csv"]
)

test_df = pd.read_csv(
    CONFIG["test_csv"]
)

sample_sub = pd.read_csv(
    CONFIG["sample_submission"]
)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Sample submission shape:", sample_sub.shape)

print("\nTrain columns:")
print(train_df.columns.tolist())

print("\nTest columns:")
print(test_df.columns.tolist())


# ============================================================
# 5. NORMALIZE COLUMN NAMES
# ============================================================

# Train
if len(train_df.columns) >= 2:

    train_df = train_df.rename(
        columns={
            train_df.columns[0]: "filename",
            train_df.columns[1]: "score"
        }
    )


# Test
test_df = test_df.rename(
    columns={
        test_df.columns[0]: "filename"
    }
)

print("\nNormalized columns:")
print(train_df.columns.tolist())
print(test_df.columns.tolist())


# ============================================================
# 6. DATA AUDIT
# ============================================================

print("\n========== DATA AUDIT ==========")

print("Training samples:", len(train_df))
print("Testing samples:", len(test_df))

print(
    "Training audio files:",
    len([
        x for x in os.listdir(
            CONFIG["train_audio_dir"]
        )
        if x.endswith(".wav")
    ])
)

print(
    "Testing audio files:",
    len([
        x for x in os.listdir(
            CONFIG["test_audio_dir"]
        )
        if x.endswith(".wav")
    ])
)

print("\nScore distribution:")
print(
    train_df["score"]
    .value_counts()
    .sort_index()
)

print("\nScore statistics:")
print(
    train_df["score"].describe()
)


# ============================================================
# 7. WHISPER TRANSCRIPTION
# ============================================================

print("\n========== LOADING WHISPER ==========")

asr_model = whisper.load_model(
    CONFIG["whisper_model"],
    device=device
)

print("Whisper loaded.")


# ------------------------------------------------------------
# Cache locations
# ------------------------------------------------------------

TRAIN_TRANSCRIPT_CACHE = (
    "/kaggle/working/train_transcripts.json"
)

TEST_TRANSCRIPT_CACHE = (
    "/kaggle/working/test_transcripts.json"
)


# ------------------------------------------------------------
# Transcription function
# ------------------------------------------------------------

def transcribe_audio_dir(
    df,
    audio_dir,
    cache_path
):

    # Load cache if available
    if os.path.exists(cache_path):

        print(
            "Loading transcript cache:",
            cache_path
        )

        with open(
            cache_path,
            "r",
            encoding="utf-8"
        ) as f:

            return json.load(f)


    results = {}

    for filename in tqdm(
        df["filename"],
        desc=f"Whisper: {os.path.basename(audio_dir)}"
    ):

        original_filename = filename

        if not str(filename).lower().endswith(".wav"):
            filename = f"{filename}.wav"

        filepath = os.path.join(
            audio_dir,
            filename
        )

        if not os.path.exists(filepath):

            results[original_filename] = {
                "text": "",
                "segments": []
            }

            continue


        try:

            result = asr_model.transcribe(

                filepath,

                language="en",

                temperature=0,

                word_timestamps=True,

                condition_on_previous_text=False

            )

            results[original_filename] = {

                "text":
                    result.get(
                        "text",
                        ""
                    ).strip(),

                "segments":
                    result.get(
                        "segments",
                        []
                    )

            }

        except Exception as e:

            print(
                f"Whisper error for {filename}: {e}"
            )

            results[original_filename] = {
                "text": "",
                "segments": []
            }


    with open(
        cache_path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            results,
            f
        )

    return results


# ------------------------------------------------------------
# Run transcription
# ------------------------------------------------------------

train_transcripts = transcribe_audio_dir(
    train_df,
    CONFIG["train_audio_dir"],
    TRAIN_TRANSCRIPT_CACHE
)

test_transcripts = transcribe_audio_dir(
    test_df,
    CONFIG["test_audio_dir"],
    TEST_TRANSCRIPT_CACHE
)


# ============================================================
# 8. ADD TRANSCRIPTS TO DATAFRAMES
# ============================================================

train_df["text"] = [
    train_transcripts.get(
        filename,
        {}
    ).get(
        "text",
        ""
    )
    for filename in train_df["filename"]
]


test_df["text"] = [
    test_transcripts.get(
        filename,
        {}
    ).get(
        "text",
        ""
    )
    for filename in test_df["filename"]
]


print("\nSample transcripts:")
display(
    train_df[
        ["filename", "score", "text"]
    ].head(10)
)


# ============================================================
# 9. WHISPER TIMING / CONFIDENCE FEATURES
# ============================================================

def extract_whisper_features(
    transcript
):

    segments = transcript.get(
        "segments",
        []
    )

    if not segments:

        return {

            "speech_seconds": 0.0,
            "span_seconds": 0.0,
            "pause_mean": 0.0,
            "pause_std": 0.0,
            "pause_p50": 0.0,
            "pause_p90": 0.0,
            "long_pause_rate": 0.0,
            "very_long_pause_rate": 0.0,
            "segment_count": 0.0,
            "asr_avg_logprob": 0.0,
            "asr_no_speech_prob": 0.0,
            "asr_compression_ratio": 0.0
        }


    durations = np.array([

        max(
            0.0,
            float(
                s.get("end", 0)
            )
            -
            float(
                s.get("start", 0)
            )
        )

        for s in segments

    ])


    gaps = np.array([

        max(
            0.0,

            float(
                segments[i + 1].get(
                    "start",
                    0
                )
            )

            -

            float(
                segments[i].get(
                    "end",
                    0
                )
            )

        )

        for i in range(
            len(segments) - 1
        )

    ])


    speech_seconds = float(
        durations.sum()
    )


    span_seconds = float(

        float(
            segments[-1].get(
                "end",
                0
            )
        )

        -

        float(
            segments[0].get(
                "start",
                0
            )
        )

    )


    if len(gaps):

        pause_mean = float(
            np.mean(gaps)
        )

        pause_std = float(
            np.std(gaps)
        )

        pause_p50 = float(
            np.percentile(
                gaps,
                50
            )
        )

        pause_p90 = float(
            np.percentile(
                gaps,
                90
            )
        )

        long_pause_rate = float(
            np.mean(gaps >= 1.0)
        )

        very_long_pause_rate = float(
            np.mean(gaps >= 2.0)
        )

    else:

        pause_mean = 0
        pause_std = 0
        pause_p50 = 0
        pause_p90 = 0
        long_pause_rate = 0
        very_long_pause_rate = 0


    weights = (

        durations

        if durations.sum() > 0

        else np.ones(
            len(durations)
        )
    )


    logprobs = np.array([

        float(
            s.get(
                "avg_logprob",
                0
            )
        )

        for s in segments

    ])


    no_speech = np.array([

        float(
            s.get(
                "no_speech_prob",
                0
            )
        )

        for s in segments

    ])


    compression = np.array([

        float(
            s.get(
                "compression_ratio",
                0
            )
        )

        for s in segments

    ])


    return {

        "speech_seconds":
            speech_seconds,

        "span_seconds":
            span_seconds,

        "pause_mean":
            pause_mean,

        "pause_std":
            pause_std,

        "pause_p50":
            pause_p50,

        "pause_p90":
            pause_p90,

        "long_pause_rate":
            long_pause_rate,

        "very_long_pause_rate":
            very_long_pause_rate,

        "segment_count":
            float(len(segments)),

        "asr_avg_logprob":
            float(
                np.average(
                    logprobs,
                    weights=weights
                )
            ),

        "asr_no_speech_prob":
            float(
                np.average(
                    no_speech,
                    weights=weights
                )
            ),

        "asr_compression_ratio":
            float(
                np.average(
                    compression,
                    weights=weights
                )
            )
    }


def build_whisper_features(
    df,
    transcript_dict
):

    rows = []

    for filename in df["filename"]:

        transcript = transcript_dict.get(
            filename,
            {}
        )

        rows.append(
            extract_whisper_features(
                transcript
            )
        )

    return pd.DataFrame(rows)


whisper_train = build_whisper_features(
    train_df,
    train_transcripts
)

whisper_test = build_whisper_features(
    test_df,
    test_transcripts
)

print(
    "Whisper feature shape:",
    whisper_train.shape
)


# ============================================================
# 10. ACOUSTIC FEATURES
# ============================================================

def summarize_feature(
    prefix,
    values
):

    values = np.asarray(
        values,
        dtype=np.float64
    )

    values = values[
        np.isfinite(values)
    ]

    if len(values) == 0:

        return {

            f"{prefix}_mean": 0.0,
            f"{prefix}_std": 0.0,
            f"{prefix}_p10": 0.0,
            f"{prefix}_p50": 0.0,
            f"{prefix}_p90": 0.0

        }


    return {

        f"{prefix}_mean":
            float(
                np.mean(values)
            ),

        f"{prefix}_std":
            float(
                np.std(values)
            ),

        f"{prefix}_p10":
            float(
                np.percentile(
                    values,
                    10
                )
            ),

        f"{prefix}_p50":
            float(
                np.percentile(
                    values,
                    50
                )
            ),

        f"{prefix}_p90":
            float(
                np.percentile(
                    values,
                    90
                )
            )
    }


def extract_acoustic_features(
    filepath
):

    try:

        y, sr = librosa.load(
            filepath,
            sr=16000,
            mono=True
        )

    except Exception:

        return {}


    if len(y) == 0:

        return {}


    duration = (
        len(y) / sr
    )


    frame_length = 1024
    hop_length = 512


    # --------------------------------------------------------
    # RMS
    # --------------------------------------------------------

    rms = librosa.feature.rms(
        y=y,
        frame_length=frame_length,
        hop_length=hop_length
    )[0]


    # --------------------------------------------------------
    # Zero crossing rate
    # --------------------------------------------------------

    zcr = librosa.feature.zero_crossing_rate(
        y,
        frame_length=frame_length,
        hop_length=hop_length
    )[0]


    # --------------------------------------------------------
    # STFT
    # --------------------------------------------------------

    spectrum = np.abs(

        librosa.stft(

            y,

            n_fft=frame_length,

            hop_length=hop_length

        )

    )


    # --------------------------------------------------------
    # Spectral features
    # --------------------------------------------------------

    centroid = librosa.feature.spectral_centroid(
        S=spectrum,
        sr=sr
    )[0]


    bandwidth = librosa.feature.spectral_bandwidth(
        S=spectrum,
        sr=sr
    )[0]


    rolloff = librosa.feature.spectral_rolloff(
        S=spectrum,
        sr=sr,
        roll_percent=0.85
    )[0]


    flatness = librosa.feature.spectral_flatness(
        S=spectrum
    )[0]


    # --------------------------------------------------------
    # MFCC
    # --------------------------------------------------------

    mfcc = librosa.feature.mfcc(

        y=y,

        sr=sr,

        n_mfcc=20,

        n_fft=frame_length,

        hop_length=hop_length

    )


    # --------------------------------------------------------
    # Voice activity
    # --------------------------------------------------------

    positive_rms = rms[
        rms > 0
    ]


    if len(positive_rms):

        noise_floor = np.percentile(
            positive_rms,
            10
        )

    else:

        noise_floor = 0.0


    active_threshold = max(
        noise_floor * 2.5,
        1e-4
    )


    active = (
        rms > active_threshold
    )


    silence_ratio = float(
        np.mean(~active)
    )


    transitions = np.diff(
        active.astype(np.int8),
        prepend=0
    )


    speech_segment_rate = (

        float(
            np.sum(
                transitions == 1
            )
        )

        /

        max(
            duration,
            1e-6
        )

    )


    # --------------------------------------------------------
    # Pitch
    # --------------------------------------------------------

    try:

        f0 = librosa.yin(

            y,

            fmin=65,

            fmax=300,

            sr=sr

        )

        f0 = f0[
            np.isfinite(f0)
            &
            (f0 > 0)
        ]

    except Exception:

        f0 = np.array([])


    # --------------------------------------------------------
    # Base features
    # --------------------------------------------------------

    features = {

        "duration":
            float(duration),

        "sample_rate":
            float(sr),

        "peak":
            float(
                np.max(
                    np.abs(y)
                )
            ),

        "dc_offset":
            float(
                np.mean(y)
            ),

        "silence_ratio":
            silence_ratio,

        "speech_segment_rate":
            speech_segment_rate,

        "clip_ratio":
            float(
                np.mean(
                    np.abs(y) >= 0.999
                )
            )

    }


    # --------------------------------------------------------
    # Distribution statistics
    # --------------------------------------------------------

    for name, values in [

        ("rms", rms),

        ("zcr", zcr),

        ("centroid", centroid),

        ("bandwidth", bandwidth),

        ("rolloff", rolloff),

        ("flatness", flatness),

        ("pitch", f0)

    ]:

        features.update(

            summarize_feature(
                name,
                values
            )

        )


    # --------------------------------------------------------
    # MFCC statistics
    # --------------------------------------------------------

    for i, coefficient in enumerate(mfcc):

        features.update(

            summarize_feature(

                f"mfcc_{i:02d}",

                coefficient

            )

        )


    return features


def build_acoustic_features(
    df,
    audio_dir
):

    rows = []

    for filename in tqdm(
        df["filename"],
        desc=f"Acoustic: {os.path.basename(audio_dir)}"
    ):

        if not str(filename).lower().endswith(".wav"):

            filename = (
                f"{filename}.wav"
            )


        filepath = os.path.join(
            audio_dir,
            filename
        )


        features = extract_acoustic_features(
            filepath
        )


        rows.append(
            features
        )


    return (
        pd.DataFrame(rows)
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .fillna(0)
    )


acoustic_train = build_acoustic_features(
    train_df,
    CONFIG["train_audio_dir"]
)

acoustic_test = build_acoustic_features(
    test_df,
    CONFIG["test_audio_dir"]
)

print(
    "Acoustic feature shape:",
    acoustic_train.shape
)


# ============================================================
# 11. LANGUAGE / GRAMMAR FEATURES
# ============================================================

print(
    "\n========== LANGUAGE TOOL =========="
)

grammar_tool = language_tool_python.LanguageTool(
    "en-US"
)


WORD_RE = re.compile(
    r"[A-Za-z]+(?:'[A-Za-z]+)?"
)


FILLERS = {

    "uh",
    "um",
    "erm",
    "hmm",
    "like",
    "basically",
    "actually",
    "literally"

}


SUBORDINATORS = {

    "although",
    "because",
    "before",
    "despite",
    "if",
    "once",
    "since",
    "though",
    "unless",
    "until",
    "when",
    "whenever",
    "whereas",
    "while",
    "whoever",
    "which",
    "that"

}


COORDINATORS = {

    "and",
    "but",
    "for",
    "nor",
    "or",
    "so",
    "yet"

}


DISCOURSE = {

    "however",
    "therefore",
    "moreover",
    "furthermore",
    "nevertheless",
    "consequently"

}


def mattr(
    tokens,
    window=50
):

    if not tokens:

        return 0.0


    if len(tokens) <= window:

        return (
            len(set(tokens))
            /
            len(tokens)
        )


    values = [

        len(
            set(
                tokens[i:i+window]
            )
        )
        /
        window

        for i in range(
            len(tokens) - window + 1
        )

    ]

    return float(
        np.mean(values)
    )


def safe_textstat(
    function,
    text
):

    try:

        value = float(
            function(text)
        )

        if np.isfinite(value):

            return value

    except Exception:

        pass

    return 0.0


def extract_grammar_features(
    text
):

    text = str(text).strip()


    tokens = [

        token.lower()

        for token in
        WORD_RE.findall(text)

    ]


    word_count = len(tokens)

    counts = Counter(tokens)


    if word_count == 0:

        return {

            "word_count": 0,
            "sentence_count": 0,
            "words_per_sentence": 0,
            "characters_per_word": 0,
            "unique_word_ratio": 0,
            "hapax_ratio": 0,
            "mattr_50": 0,
            "filler_rate_100": 0,
            "subordinator_rate_100": 0,
            "coordinator_rate_100": 0,
            "discourse_rate_100": 0,
            "adjacent_repeat_rate_100": 0,
            "grammar_errors_100": 0,
            "typos_100": 0,
            "style_errors_100": 0,
            "all_languagetool_issues_100": 0,
            "unique_grammar_rules": 0,
            "misspelling_issues_100": 0,
            "readability_flesch": 0,
            "readability_grade": 0,
            "difficult_words_100": 0

        }


    # --------------------------------------------------------
    # Sentences
    # --------------------------------------------------------

    sentence_count = max(

        1,

        len(
            [
                s
                for s in re.split(
                    r"[.!?]+",
                    text
                )
                if s.strip()
            ]
        )

    )


    # --------------------------------------------------------
    # LanguageTool
    # --------------------------------------------------------

    try:

        matches = grammar_tool.check(
            text
        )

    except Exception:

        matches = []


    category_counts = Counter(

        str(
            match.category
        ).upper()

        for match in matches

    )


    issue_types = Counter(

        str(
            match.rule_issue_type
        ).upper()

        for match in matches

    )


    rule_ids = set(

        str(
            match.rule_id
        )

        for match in matches

    )


    grammar_errors = (
        category_counts["GRAMMAR"]
    )


    typo_errors = (
        category_counts["TYPOS"]
        +
        issue_types["MISSPELLING"]
    )


    style_errors = (
        category_counts["STYLE"]
    )


    # --------------------------------------------------------
    # Lexical
    # --------------------------------------------------------

    unique_words = len(
        set(tokens)
    )


    unique_word_ratio = (
        unique_words
        /
        max(
            word_count,
            1
        )
    )


    hapax_ratio = (

        sum(
            value == 1
            for value in counts.values()
        )

        /

        max(
            word_count,
            1
        )

    )


    avg_word_len = (

        np.mean(
            [
                len(w)
                for w in tokens
            ]
        )

        if tokens

        else 0

    )


    words_per_sentence = (

        word_count
        /
        sentence_count

    )


    # --------------------------------------------------------
    # Repetition
    # --------------------------------------------------------

    adjacent_repeats = sum(

        a == b

        for a, b in zip(
            tokens,
            tokens[1:]
        )

    )


    # --------------------------------------------------------
    # Linguistic complexity
    # --------------------------------------------------------

    filler_count = sum(

        counts[word]

        for word in FILLERS

    )


    subordinator_count = sum(

        counts[word]

        for word in SUBORDINATORS

    )


    coordinator_count = sum(

        counts[word]

        for word in COORDINATORS

    )


    discourse_count = sum(

        counts[word]

        for word in DISCOURSE

    )


    denominator = max(
        word_count,
        1
    )


    return {

        "word_count":
            float(word_count),

        "sentence_count":
            float(sentence_count),

        "words_per_sentence":
            float(words_per_sentence),

        "characters_per_word":
            float(avg_word_len),

        "unique_word_ratio":
            float(unique_word_ratio),

        "hapax_ratio":
            float(hapax_ratio),

        "mattr_50":
            float(
                mattr(
                    tokens,
                    50
                )
            ),

        "filler_rate_100":
            100
            *
            filler_count
            /
            denominator,

        "subordinator_rate_100":
            100
            *
            subordinator_count
            /
            denominator,

        "coordinator_rate_100":
            100
            *
            coordinator_count
            /
            denominator,

        "discourse_rate_100":
            100
            *
            discourse_count
            /
            denominator,

        "adjacent_repeat_rate_100":
            100
            *
            adjacent_repeats
            /
            denominator,

        "grammar_errors_100":
            100
            *
            grammar_errors
            /
            denominator,

        "typos_100":
            100
            *
            typo_errors
            /
            denominator,

        "style_errors_100":
            100
            *
            style_errors
            /
            denominator,

        "all_languagetool_issues_100":
            100
            *
            len(matches)
            /
            denominator,

        "unique_grammar_rules":
            float(
                len(rule_ids)
            ),

        "misspelling_issues_100":
            100
            *
            issue_types["MISSPELLING"]
            /
            denominator,

        "readability_flesch":
            safe_textstat(
                textstat.flesch_reading_ease,
                text
            ),

        "readability_grade":
            safe_textstat(
                textstat.flesch_kincaid_grade,
                text
            ),

        "difficult_words_100":
            100
            *
            safe_textstat(
                textstat.difficult_words,
                text
            )
            /
            denominator

    }


def build_grammar_features(
    df
):

    rows = []

    for text in tqdm(
        df["text"],
        desc="Grammar features"
    ):

        rows.append(
            extract_grammar_features(
                text
            )
        )

    return pd.DataFrame(
        rows
    )


grammar_train = build_grammar_features(
    train_df
)

grammar_test = build_grammar_features(
    test_df
)

print(
    "Grammar feature shape:",
    grammar_train.shape
)


# ============================================================
# 12. DISPLAY CORRELATIONS
# ============================================================

feature_target_corr = pd.DataFrame({

    "feature":
        grammar_train.columns,

    "correlation": [

        np.corrcoef(
            grammar_train[col].values,
            train_df["score"].values
        )[0, 1]

        if grammar_train[col].std() > 0

        else 0

        for col in grammar_train.columns

    ]

})


print(
    "\nTop grammar feature correlations:"
)

display(
    feature_target_corr
    .sort_values(
        "correlation",
        key=lambda x: x.abs(),
        ascending=False
    )
    .head(15)
)


# ============================================================
# 13. COMBINE NUMERIC FEATURES
# ============================================================

numeric_train = pd.concat(

    [

        acoustic_train.reset_index(
            drop=True
        ),

        whisper_train.reset_index(
            drop=True
        ),

        grammar_train.reset_index(
            drop=True
        )

    ],

    axis=1

)


numeric_test = pd.concat(

    [

        acoustic_test.reset_index(
            drop=True
        ),

        whisper_test.reset_index(
            drop=True
        ),

        grammar_test.reset_index(
            drop=True
        )

    ],

    axis=1

)


numeric_train = (

    numeric_train

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


numeric_test = (

    numeric_test

    .replace(
        [np.inf, -np.inf],
        np.nan
    )

    .fillna(0)

)


X_numeric_train = (
    numeric_train.values
)

X_numeric_test = (
    numeric_test.values
)


print(
    "\nTotal engineered numeric features:",
    X_numeric_train.shape[1]
)


# ============================================================
# 14. MPNet EMBEDDINGS
# ============================================================

print(
    "\n========== MPNet EMBEDDINGS =========="
)

text_encoder = SentenceTransformer(

    CONFIG["embedding_model"],

    device=device

)


emb_train = text_encoder.encode(

    train_df["text"].fillna("").tolist(),

    batch_size=16,

    show_progress_bar=True,

    normalize_embeddings=True,

    convert_to_numpy=True

)


emb_test = text_encoder.encode(

    test_df["text"].fillna("").tolist(),

    batch_size=16,

    show_progress_bar=True,

    normalize_embeddings=True,

    convert_to_numpy=True

)


print(
    "Embedding shape:",
    emb_train.shape
)


# ============================================================
# 15. FUSED NUMERIC + MPNET
# ============================================================

X_fused_train = np.hstack(

    [

        X_numeric_train,

        emb_train

    ]

)


X_fused_test = np.hstack(

    [

        X_numeric_test,

        emb_test

    ]

)


# ============================================================
# 16. TARGET
# ============================================================

y = (
    train_df["score"]
    .astype(float)
    .values
)


# ============================================================
# 17. STRATIFIED CROSS VALIDATION
# ============================================================

# Scores are discrete / ordinal.
# We preserve their distribution across folds.

strata = np.where(

    y < 2.0,

    "under_2",

    y.astype(str)

)


skf = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=SEED

)


folds = list(

    skf.split(
        np.zeros(len(y)),
        strata
    )

)


print(
    "\nFold sizes:"
)

for i, (tr, va) in enumerate(
    folds,
    1
):

    print(
        f"Fold {i}:",
        len(tr),
        "train /",
        len(va),
        "validation"
    )


# ============================================================
# 18. METRIC FUNCTION
# ============================================================

def calculate_metrics(
    y_true,
    y_pred
):

    rmse = np.sqrt(

        mean_squared_error(
            y_true,
            y_pred
        )

    )

    pearson = pearsonr(

        y_true,
        y_pred

    )[0]

    return rmse, pearson


# ============================================================
# 19. STORAGE FOR OOF PREDICTIONS
# ============================================================

oof_predictions = {}

test_predictions = {}


# ============================================================
# 20. EXTRA TREES
# ============================================================

print(
    "\n========== EXTRA TREES =========="
)

extra_oof = np.zeros(
    len(y)
)

extra_test_folds = []


for fold_id, (
    train_idx,
    val_idx
) in enumerate(
    folds,
    1
):

    model = ExtraTreesRegressor(

        n_estimators=700,

        min_samples_leaf=3,

        max_features=0.72,

        n_jobs=-1,

        random_state=
            SEED + fold_id

    )


    model.fit(

        X_numeric_train[
            train_idx
        ],

        y[
            train_idx
        ]

    )


    extra_oof[
        val_idx
    ] = model.predict(

        X_numeric_train[
            val_idx
        ]

    )


    extra_test_folds.append(

        model.predict(
            X_numeric_test
        )

    )


extra_test = np.mean(
    extra_test_folds,
    axis=0
)


oof_predictions[
    "extra_trees"
] = extra_oof


test_predictions[
    "extra_trees"
] = extra_test


rmse, pearson = calculate_metrics(
    y,
    extra_oof
)


print(
    f"ExtraTrees RMSE: {rmse:.5f}"
)

print(
    f"ExtraTrees Pearson: {pearson:.5f}"
)


# ============================================================
# 21. LIGHTGBM
# ============================================================

print(
    "\n========== LIGHTGBM =========="
)

lgb_oof = np.zeros(
    len(y)
)

lgb_test_folds = []


for fold_id, (
    train_idx,
    val_idx
) in enumerate(
    folds,
    1
):

    model = LGBMRegressor(

        n_estimators=650,

        learning_rate=0.025,

        num_leaves=12,

        max_depth=5,

        min_child_samples=24,

        subsample=0.85,

        colsample_bytree=0.72,

        reg_lambda=8.0,

        verbosity=-1,

        random_state=
            SEED + fold_id,

        n_jobs=-1

    )


    model.fit(

        X_numeric_train[
            train_idx
        ],

        y[
            train_idx
        ]

    )


    lgb_oof[
        val_idx
    ] = model.predict(

        X_numeric_train[
            val_idx
        ]

    )


    lgb_test_folds.append(

        model.predict(
            X_numeric_test
        )

    )


lgb_test = np.mean(
    lgb_test_folds,
    axis=0
)


oof_predictions[
    "lightgbm"
] = lgb_oof


test_predictions[
    "lightgbm"
] = lgb_test


rmse, pearson = calculate_metrics(
    y,
    lgb_oof
)


print(
    f"LightGBM RMSE: {rmse:.5f}"
)

print(
    f"LightGBM Pearson: {pearson:.5f}"
)


# ============================================================
# 22. MPNet + NUMERIC RIDGE
# ============================================================

print(
    "\n========== MPNet + NUMERIC RIDGE =========="
)

ridge_oof = np.zeros(
    len(y)
)

ridge_test_folds = []


for fold_id, (
    train_idx,
    val_idx
) in enumerate(
    folds,
    1
):

    model = make_pipeline(

        StandardScaler(),

        Ridge(
            alpha=120.0
        )

    )


    model.fit(

        X_fused_train[
            train_idx
        ],

        y[
            train_idx
        ]

    )


    ridge_oof[
        val_idx
    ] = model.predict(

        X_fused_train[
            val_idx
        ]

    )


    ridge_test_folds.append(

        model.predict(
            X_fused_test
        )

    )


ridge_test = np.mean(
    ridge_test_folds,
    axis=0
)


oof_predictions[
    "mpnet_ridge"
] = ridge_oof


test_predictions[
    "mpnet_ridge"
] = ridge_test


rmse, pearson = calculate_metrics(
    y,
    ridge_oof
)


print(
    f"MPNet Ridge RMSE: {rmse:.5f}"
)

print(
    f"MPNet Ridge Pearson: {pearson:.5f}"
)


# ============================================================
# 23. TF-IDF WORD + CHARACTER RIDGE
# ============================================================

print(
    "\n========== TF-IDF RIDGE =========="
)

tfidf_oof = np.zeros(
    len(y)
)

tfidf_test_folds = []


for fold_id, (
    train_idx,
    val_idx
) in enumerate(
    folds,
    1
):

    print(
        f"TF-IDF fold {fold_id}"
    )


    vectorizer = FeatureUnion(

        [

            (

                "word",

                TfidfVectorizer(

                    ngram_range=(1, 2),

                    min_df=2,

                    max_features=30000,

                    sublinear_tf=True,

                    strip_accents="unicode"

                )

            ),

            (

                "char",

                TfidfVectorizer(

                    analyzer="char_wb",

                    ngram_range=(3, 5),

                    min_df=3,

                    max_features=30000,

                    sublinear_tf=True

                )

            )

        ],

        transformer_weights={

            "word": 1.0,

            "char": 0.55

        }

    )


    Xtr = vectorizer.fit_transform(

        train_df[
            "text"
        ]
        .iloc[
            train_idx
        ]

    )


    Xval = vectorizer.transform(

        train_df[
            "text"
        ]
        .iloc[
            val_idx
        ]

    )


    Xtest = vectorizer.transform(

        test_df[
            "text"
        ]

    )


    model = Ridge(

        alpha=18.0,

        solver="lsqr"

    )


    model.fit(

        Xtr,

        y[
            train_idx
        ]

    )


    tfidf_oof[
        val_idx
    ] = model.predict(
        Xval
    )


    tfidf_test_folds.append(

        model.predict(
            Xtest
        )

    )


tfidf_test = np.mean(
    tfidf_test_folds,
    axis=0
)


oof_predictions[
    "tfidf_ridge"
] = tfidf_oof


test_predictions[
    "tfidf_ridge"
] = tfidf_test


rmse, pearson = calculate_metrics(
    y,
    tfidf_oof
)


print(
    f"TF-IDF Ridge RMSE: {rmse:.5f}"
)

print(
    f"TF-IDF Ridge Pearson: {pearson:.5f}"
)


# ============================================================
# 24. BASE MODEL COMPARISON
# ============================================================

model_names = list(
    oof_predictions.keys()
)


metrics_rows = []


for name in model_names:

    rmse, pearson = calculate_metrics(

        y,

        oof_predictions[name]

    )


    metrics_rows.append({

        "model":
            name,

        "RMSE":
            rmse,

        "Pearson":
            pearson

    })


metrics_df = pd.DataFrame(
    metrics_rows
)


print(
    "\n========== BASE MODEL PERFORMANCE =========="
)

display(
    metrics_df.sort_values(
        "RMSE"
    )
)


# ============================================================
# 25. OOF CORRELATION BETWEEN MODELS
# ============================================================

oof_matrix = np.column_stack(

    [

        oof_predictions[name]

        for name in model_names

    ]

)


test_matrix = np.column_stack(

    [

        test_predictions[name]

        for name in model_names

    ]

)


corr_df = pd.DataFrame(

    oof_matrix,

    columns=model_names

)


plt.figure(
    figsize=(9, 7)
)

sns.heatmap(

    corr_df.corr(),

    annot=True,

    fmt=".3f",

    cmap="coolwarm",

    center=0

)

plt.title(
    "OOF Prediction Correlation"
)

plt.tight_layout()

plt.show()


# ============================================================
# 26. OPTIMIZE ENSEMBLE WEIGHTS
# ============================================================

print(
    "\n========== OPTIMIZING ENSEMBLE =========="
)


def ensemble_objective(
    weights
):

    prediction = (
        oof_matrix @ weights
    )

    return np.mean(
        (
            y - prediction
        ) ** 2
    )


initial_weights = (

    np.ones(
        len(model_names)
    )

    /

    len(model_names)

)


result = minimize(

    ensemble_objective,

    initial_weights,

    bounds=[
        (0.0, 1.0)
        for _ in model_names
    ],

    constraints={

        "type": "eq",

        "fun":
            lambda w:
            np.sum(w) - 1.0

    },

    method="SLSQP",

    options={

        "ftol": 1e-12,

        "maxiter": 5000

    }

)


if not result.success:

    print(
        "Optimization warning:",
        result.message
    )


weights = result.x


weights_df = pd.DataFrame({

    "model":
        model_names,

    "weight":
        weights

})


print(
    "\nOptimized weights:"
)

display(
    weights_df
    .sort_values(
        "weight",
        ascending=False
    )
)


# ============================================================
# 27. FINAL OOF PREDICTION
# ============================================================

final_oof = (

    oof_matrix

    @

    weights

)


final_test = (

    test_matrix

    @

    weights

)


# ------------------------------------------------------------
# Calibration / bounds
# ------------------------------------------------------------

final_oof = np.clip(
    final_oof,
    0.0,
    5.0
)


final_test = np.clip(
    final_test,
    0.0,
    5.0
)


# ============================================================
# 28. MANDATORY TRAINING RMSE
# ============================================================

training_rmse = np.sqrt(

    mean_squared_error(
        y,
        final_oof
    )

)


training_pearson = pearsonr(

    y,
    final_oof

)[0]


print(
    "\n"
    "================================================"
)

print(
    "MANDATORY TRAINING-DATA OOF RMSE"
)

print(
    "================================================"
)

print(
    f"Training / OOF RMSE: "
    f"{training_rmse:.6f}"
)

print(
    f"Training / OOF Pearson: "
    f"{training_pearson:.6f}"
)

print(
    "================================================"
)


# ============================================================
# 29. MODEL + ENSEMBLE METRICS
# ============================================================

ensemble_metrics = metrics_df.copy()


ensemble_metrics = pd.concat(

    [

        ensemble_metrics,

        pd.DataFrame([{

            "model":
                "OPTIMIZED ENSEMBLE",

            "RMSE":
                training_rmse,

            "Pearson":
                training_pearson

        }])

    ],

    ignore_index=True

)


print(
    "\n========== FINAL METRICS =========="
)

display(
    ensemble_metrics.sort_values(
        "RMSE"
    )
)


# ============================================================
# 30. ACTUAL VS PREDICTED
# ============================================================

plt.figure(
    figsize=(9, 8)
)

sns.scatterplot(

    x=y,

    y=final_oof,

    alpha=0.70,

    s=70

)


plt.plot(

    [0, 5],

    [0, 5],

    "--",

    linewidth=2

)


plt.xlabel(
    "Ground Truth Grammar Score"
)

plt.ylabel(
    "OOF Predicted Grammar Score"
)

plt.title(

    f"OOF Predictions vs Ground Truth\n"
    f"RMSE = {training_rmse:.4f} | "
    f"Pearson = {training_pearson:.4f}"

)

plt.xlim(
    0.8,
    5.2
)

plt.ylim(
    0.8,
    5.2
)

plt.tight_layout()

plt.show()


# ============================================================
# 31. RESIDUAL ANALYSIS
# ============================================================

residuals = (
    y - final_oof
)


plt.figure(
    figsize=(10, 6)
)

sns.histplot(

    residuals,

    bins=25,

    kde=True

)

plt.axvline(

    0,

    linestyle="--",

    linewidth=2

)

plt.xlabel(
    "Residual = Actual - Prediction"
)

plt.title(
    "OOF Residual Distribution"
)

plt.tight_layout()

plt.show()


# ============================================================
# 32. RMSE BY SCORE
# ============================================================

score_analysis = pd.DataFrame({

    "actual":
        y,

    "prediction":
        final_oof,

    "residual":
        residuals

})


score_analysis["abs_error"] = (

    score_analysis["residual"]
    .abs()

)


score_group = (

    score_analysis

    .groupby("actual")

    .agg(

        count=("actual", "size"),

        mean_prediction=(
            "prediction",
            "mean"
        ),

        rmse=(
            "residual",
            lambda x:
            np.sqrt(
                np.mean(
                    x ** 2
                )
            )
        ),

        mae=(
            "abs_error",
            "mean"
        )

    )

    .reset_index()

)


print(
    "\n========== PERFORMANCE BY SCORE =========="
)

display(
    score_group
)


# ============================================================
# 33. RESIDUAL VS PREDICTION
# ============================================================

plt.figure(
    figsize=(10, 6)
)

sns.scatterplot(

    x=final_oof,

    y=residuals,

    alpha=0.65

)

plt.axhline(

    0,

    linestyle="--",

    linewidth=2

)

plt.xlabel(
    "Predicted Score"
)

plt.ylabel(
    "Residual"
)

plt.title(
    "OOF Residual vs Predicted Score"
)

plt.tight_layout()

plt.show()


# ============================================================
# 34. CHECK PREDICTION DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10, 6)
)

sns.histplot(

    y,

    bins=20,

    kde=True,

    stat="density",

    alpha=0.35,

    label="Actual"

)

sns.histplot(

    final_oof,

    bins=20,

    kde=True,

    stat="density",

    alpha=0.35,

    label="OOF Prediction"

)

plt.xlabel(
    "Grammar Score"
)

plt.title(
    "Actual vs OOF Prediction Distribution"
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 35. TEST PREDICTION SANITY CHECK
# ============================================================

print(
    "\n========== TEST PREDICTION DISTRIBUTION =========="
)

print(
    pd.Series(
        final_test
    ).describe()
)


print(
    "\nMin:",
    final_test.min()
)

print(
    "Max:",
    final_test.max()
)

print(
    "Mean:",
    final_test.mean()
)

print(
    "Std:",
    final_test.std()
)


# ============================================================
# 36. SAVE MODEL ARTIFACTS
# ============================================================

artifact_dir = (
    "/kaggle/working/grammar_artifacts"
)

os.makedirs(
    artifact_dir,
    exist_ok=True
)


# OOF predictions

oof_output = pd.DataFrame({

    "filename":
        train_df["filename"],

    "actual_score":
        y,

    **{
        name:
            oof_predictions[name]

        for name in model_names
    },

    "ensemble":
        final_oof

})


oof_output.to_csv(

    os.path.join(
        artifact_dir,
        "oof_predictions.csv"
    ),

    index=False

)


# Model weights

weights_df.to_csv(

    os.path.join(
        artifact_dir,
        "ensemble_weights.csv"
    ),

    index=False

)


# Metrics

ensemble_metrics.to_csv(

    os.path.join(
        artifact_dir,
        "metrics.csv"
    ),

    index=False

)


# ============================================================
# 37. CREATE FINAL SUBMISSION
# ============================================================

print(
    "\n========== CREATING SUBMISSION =========="
)


# IMPORTANT:
# Use test.csv order, not sample_submission row count,
# because the competition dataset may contain a stale
# sample_submission file.

submission = pd.DataFrame({

    "filename":
        test_df["filename"],

    "label":
        final_test

})


submission["label"] = (

    submission["label"]

    .astype(float)

    .clip(
        0,
        5
    )

)


submission_path = (
    "/kaggle/working/submission.csv"
)


submission.to_csv(

    submission_path,

    index=False

)


# ============================================================
# 38. VALIDATE SUBMISSION
# ============================================================

assert len(submission) == len(
    test_df
)


assert (
    submission["filename"].tolist()
    ==
    test_df["filename"].tolist()
)


assert (
    submission["label"].notna().all()
)


assert np.isfinite(
    submission["label"]
).all()


assert submission[
    "label"
].between(
    0,
    5
).all()


print(
    "Submission validation PASSED."
)


print(
    "\nSubmission shape:",
    submission.shape
)


print(
    "\nSubmission statistics:"
)

print(
    submission["label"].describe()
)


print(
    "\nFirst 10 predictions:"
)

display(
    submission.head(10)
)


print(
    "\nSubmission saved to:"
)

print(
    submission_path
)


# ============================================================
# 39. FINAL SUMMARY
# ============================================================

print(
    "\n"
    "=========================================================="
)

print(
    "FINAL MODEL SUMMARY"
)

print(
    "=========================================================="
)

print(
    f"Training samples       : {len(train_df)}"
)

print(
    f"Test samples           : {len(test_df)}"
)

print(
    f"Numeric features       : {X_numeric_train.shape[1]}"
)

print(
    f"MPNet dimensions       : {emb_train.shape[1]}"
)

print(
    f"Final OOF RMSE         : {training_rmse:.6f}"
)

print(
    f"Final OOF Pearson      : {training_pearson:.6f}"
)

print(
    "\nModel weights:"
)

for name, weight in zip(
    model_names,
    weights
):

    print(
        f"  {name:20s}: "
        f"{weight:.4f}"
    )

print(
    "\nSubmission:"
)

print(
    submission_path
)

print(
    "=========================================================="
)
```
